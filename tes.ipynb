```python
# Pastikan jalur relatifnya benar sesuai struktur direktori
from beat_craft_sdk.core import BeatCraftConfig, BeatCraft, TransformerConfig, CraftingTransformer
from beat_craft_sdk.config import GameGenre, GameMood, GameEmotional
from itertools import product
import pandas as pd
```

```python
# Buat instance BeatCraftConfig dengan method 'transformer'
bcconfig = TransformerConfig()

# Mengatur genre, mood, dan emosional menggunakan method transformer
# Mengubah genre ke RPG
bcconfig.set_game_genre(GameGenre.Horror)
print(f"Game Genre (Transformer): {bcconfig.get_game_genre()}")  # Harus mengembalikan genre yang benar

# Mengubah mood ke RELAXING
bcconfig.set_game_mood(GameMood.JOYFUL)
print(f"Game Mood (Transformer): {bcconfig.get_game_mood()}")  # Harus mengembalikan mood yang benar

# Mengubah emosional ke FEAR
bcconfig.set_game_emotional(GameEmotional.PEACEFUL)
print(f"Game Emotional (Transformer): {bcconfig.get_game_emotional()}")  # Harus mengembalikan emosional yang benar

sdk = BeatCraft(config=bcconfig)
```

```python
# Akses output directory
print(sdk.config.get_game_mood())  # Mengakses output_dir dari BeatCraftConfig
```

## Generate dengan inputan tunggal

```python
sdk.set_melody_engine(CraftingTransformer())
# transformer way
notes = sdk.compose_melody(game_mood=bcconfig.get_game_mood(), 
                           game_emotional=bcconfig.get_game_emotional(), 
                           game_genre=bcconfig.get_game_genre())
```

## Generate dengan inputan list

```python
# Daftar kombinasi mood, genre, dan emotional
moodList = [GameMood.EPIC, GameMood.JOYFUL, GameMood.RELAXING, 
            GameMood.SERENE, GameMood.TENSE]

genreList = [GameGenre.ACTION, GameGenre.Horror, GameGenre.Puzzle,
             GameGenre.RPG]

emoList = [GameEmotional.CHALLENGE, GameEmotional.EXCITEMENT,
           GameEmotional.FEAR, GameEmotional.PEACEFUL]

# Inisialisasi engine
sdk.set_melody_engine(CraftingTransformer())

# Buat kombinasi semua parameter
combinations = list(product(moodList, genreList, emoList))

# List untuk menyimpan hasil evaluasi
summary_results = []

for mood, genre, emotional in combinations:
    # Set nilai mood, genre, dan emotional menggunakan fungsi bcconfig
    bcconfig.set_game_mood(mood)
    bcconfig.set_game_genre(genre)
    bcconfig.set_game_emotional(emotional)

    print(f"Evaluating for Mood={mood}, Genre={genre}, Emotional={emotional}")

    try:
        # Generate dan evaluasi melody
        notes, results, summary = sdk.compose_melody(
            game_mood=bcconfig.get_game_mood(),  # Mengambil nilai mood yang telah diset
            game_emotional=bcconfig.get_game_emotional(),  # Mengambil nilai emotional yang telah diset
            game_genre=bcconfig.get_game_genre()  # Mengambil nilai genre yang telah diset
        )

        # Tambahkan hasil evaluasi ke dalam summary
        summary_entry = {
            "Mood": mood,
            "Genre": genre,
            "Emotional": emotional,
            "Avg_Smoothness": summary.get('avg_smoothness'),
            "Avg_Consonance": summary.get('avg_consonance'),
            "Avg_Rhythmic_Variety": summary.get('avg_rhythmic_variety'),
            "Avg_Range": summary.get('avg_range'),
            "Avg_Repetition_Rate": summary.get('avg_repetition_rate'),
            "Avg_Tonality": summary.get('avg_tonality'),
            "Avg_Combined_Fitness": summary.get('avg_combined_fitness'),
            "Avg_MSE_Pitch": summary.get('avg_mse_pitch'),
            "Avg_Euclidean_Distance": summary.get('avg_euclidean_distance')
        }
        
        summary_results.append(summary_entry)
    except ValueError as e:
        # Tangani kasus ketika kombinasi menghasilkan data kosong
        print(f"Skipping combination Mood={mood}, Genre={genre}, Emotional={emotional} due to error: {e}")
        continue  # Lanjutkan ke iterasi berikutnya

# Konversi hasil evaluasi menjadi DataFrame
summary_df = pd.DataFrame(summary_results)

# Simpan tabel hasil evaluasi ke file CSV
output_path = "./assets/evaluation_summary.csv"
summary_df.to_csv(output_path, index=False)
print(f"Rangkuman evaluasi disimpan di {output_path}")
```

## Conclusion

```python
# Pastikan kolom Genre, Mood, dan Emotional adalah string
summary_df['Genre'] = summary_df['Genre'].astype(str)
summary_df['Mood'] = summary_df['Mood'].astype(str)
summary_df['Emotional'] = summary_df['Emotional'].astype(str)

# Membuat DataFrame ringkasan dengan rata-rata per kombinasi Genre, Mood, dan Emotional
summary_grouped = summary_df.groupby(['Genre', 'Mood', 'Emotional']).mean().reset_index()

# Secara otomatis memilih semua kolom numerik
numeric_columns = summary_grouped.select_dtypes(include=['float64', 'int64']).columns.tolist()

# Pilih kolom untuk tampilan akhir (kombinasi key dan numeric data)
columns_to_include = ['Genre', 'Mood', 'Emotional'] + numeric_columns
summary_grouped = summary_grouped[columns_to_include]

# Urutkan berdasarkan kolom numerik pertama (Avg_Combined_Fitness) dan kedua (Avg_Euclidean_Distance) secara descending
summary_grouped = summary_grouped.sort_values(
    by=['Avg_Combined_Fitness', 'Avg_Euclidean_Distance'],  # Kolom untuk prioritas urutan
    ascending=[False, False]  # Descending
).reset_index(drop=True)

summary_grouped
```
